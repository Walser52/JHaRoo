```python
from ase import Atoms
from ase.visualize import view
from ase.build import bulk
from ase.geometry import get_distances
import numpy as np

# Step 1: Create CsPbBr3 perovskite structure
a = 6.0  # Approximate lattice constant for CsPbBr3
cs = 'Cs'
pb = 'Pb'
br = 'Br'

# Primitive cell
perovskite = Atoms([cs, pb] + [br] * 3,
                   positions=[(0, 0, 0),  # Cs position
                              (a/2, a/2, a/2),  # Pb position
                              (a/2, 0, 0),      # Br positions
                              (0, a/2, 0),
                              (0, 0, a/2)],
                   cell=[a, a, a])

# Step 2: Extract the Pb and Br atoms to form the PbBr6 octahedron
pb_position = perovskite.positions[1]
br_positions = perovskite.positions[2:]

octahedron = Atoms([pb] + [br] * 3,
                   positions=[pb_position] + br_positions.tolist(),
                   cell=[a, a, a])

# Step 3: Apply rotation to the PbBr6 octahedron
# Rotate around the z-axis
angle = 15  # Rotation angle in degrees
rotation_matrix = np.array([[np.cos(np.radians(angle)), -np.sin(np.radians(angle)), 0],
                            [np.sin(np.radians(angle)), np.cos(np.radians(angle)), 0],
                            [0, 0, 1]])

# Apply rotation to Br atoms
br_positions_tilted = (rotation_matrix @ (br_positions - pb_position).T).T + pb_position

# Update octahedron with tilted Br positions
octahedron.set_positions([pb_position] + br_positions_tilted.tolist())

# Step 4: Visualize the structures
print("Original Perovskite Structure:")
view(perovskite)

print("PbBr6 Octahedron:")
view(octahedron)

# You can now rotate and view the structures interactively.
```

```python
from ase import Atoms
from ase.visualize import view

# Step 1: Create CsPbBr3 perovskite structure
a = 6.0  # Approximate lattice constant for CsPbBr3
cs = 'Cs'
pb = 'Pb'
br = 'Br'

# Primitive cell
perovskite = Atoms([cs, pb] + [br] * 3,
                   positions=[(0, 0, 0),  # Cs position
                              (a/2, a/2, a/2),  # Pb position
                              (a/2, 0, 0),      # Br positions
                              (0, a/2, 0),
                              (0, 0, a/2)],
                   cell=[a, a, a])

# Step 2: Identify the atoms forming the PbBr6 octahedron
octahedron_indices = [1, 2, 3, 4]  # Indices of Pb and Br atoms

# Step 3: Apply rotation to the PbBr6 octahedron
# Rotate around the z-axis by 15 degrees, using the Pb atom as the center
perovskite.rotate(15, 'z', rotate_cell=False, center=perovskite.positions[1], 
                  indices=octahedron_indices)

# Step 4: Visualize the original and rotated structures
print("Rotated Perovskite Structure:")
view(perovskite)
```
