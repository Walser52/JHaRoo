```python
def _findBandgap(bands, printall = False, sort = False, threshold = 0.1, window = [], index = []):
  """
  Takes a ascending sorted list of bands minima and maxima and returns the maximum bandgap and the values at which that gap occurs.
  Args:
    bands: List of bands minima and maxima e.g. [[0,1], [1,2],[2.2,3], [3.5,4], [4,5]]
    printall: If True prints all bandgaps
    sort: If the bands are unsorted and need sorting.
    threshold: Only gaps greater than threshold are considered. 
    index: Only keeps the bands that lie between the index limits. 
    window: Only keeps the bands that overlap with the window provided. 

  Note:
    Both index and window cannot be given simulatneously
  """
#  bands = [[0,1], [1,2],[2.2,3], [3.5,4], [4,5]]
  if (len(index) > 0) and (len(window) > 0):
    raise ValueError("Error: Both index and window cannot be given at the same time.")

  if len(index) > 0: bands = bands[index[0]: index[1]]
  if len(window) > 0: bands = [band for band in bands if band[0] <= window[1] and band[1] >= window[0]]



  if sort: bands = sorted(bands, key=lambda x: x[0])



  bandgaps_pos = [] #bandgap min and max.
  bandgaps = np.empty(0)

  for (band_low, band_hi), (nextband_low, nextband_hi) in zip(bands, bands[1:]):
    if (nextband_low - band_hi ) > threshold: 
      bandgaps = np.append(bandgaps, [nextband_low - band_hi] )
      bandgaps_pos.append([band_hi, nextband_low])

  if len(bandgaps) > 0: 
    bandgap = np.max(bandgaps)
    
    pos = bandgaps.argmax()
    lumo, homo = bandgaps_pos[pos] 

  else:
    bandgaps = np.array([0])
    lumo, homo = None, None
  if printall: 
    print("Bandgaps:", bandgaps)
    print("Locations:",bandgaps_pos )

  return bandgap, (lumo, homo)
```
