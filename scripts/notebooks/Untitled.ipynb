```python
from pymatgen.core import Lattice, Structure

# Define the lattice parameters
a = 5.0  # Replace with your desired a and b value
c = 10.0  # Replace with your desired c value

# Create a tetragonal lattice
lattice = Lattice.tetragonal(a, c)

# Define the atomic positions
# Iodine atoms
iodine_positions = [[0, 0, 0.5], [0, 0, -0.5]]

# Bromine atoms
bromine_positions = [
    [0.5, 0.5, 0], [0.5, -0.5, 0], 
    [-0.5, 0.5, 0], [-0.5, -0.5, 0]
]

# Combine positions
positions = iodine_positions + bromine_positions

# Define the species
species = ["I", "I", "Br", "Br", "Br", "Br"]

# Create the structure
structure = Structure(lattice, species, positions)

# Optionally, you can print or save the structure
print(structure)
```
