```python
from jhr.plotters._base import makeplot
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
base = '../../data/boltzwann_files/CsPbBr3/'

files = {'boltzdos': {'file': 'CsPBBr3_boltzdos.dat', 'x': 0, 'y': (1)},
          'elcond':  {'file': 'CsPBBr3_elcond.dat', 'x': 0, 'y': (2)},
          'seebeck': {'file': 'CsPBBr3_seebeck.dat', 'x': 0, 'y': (2)} 
          
           }

data = {}
for key in files.keys():
  file = base + files[key]['file']
  data[key] = np.genfromtxt(file)

```

```python
for key in data.keys():
  x = files[key]['x']
  all_ys = files[key]['y']

  plt.figure(figsize = [4,4])

  plt.plot(data[key][:, x], data[key][:,all_ys], label = key)
  plt.legend()
  plt.show()
```
