```python
from jhr.plotters._base import makeplot
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
base = '../../data/boltzwann_files/CsPbBr3/'
#files = os.listdir(base)

# files = [
#  ('CsPBBr3_tdf.dat',  2),
#  ('CsPBBr3_boltzdos.dat', 2),
#  ('CsPBBr3_seebeck.dat',  2),
#  ('CsPBBr3_sigmas.dat', 2),
#  ('CsPBBr3_kappa.dat', 1)]


files = {'boltzdos': {'file': 'CsPBBr3_boltzdos.dat', 'x': 0, 'y': (2)},
          'elcond': {'file': 'CsPBBr3_elcond.dat', 'x': 0, 'y': (2)},
          'seebeck': {'file': 'CsPBBr3_seebeck.dat', 'x': 0, 'y': (2,3)} 
          
           }

data = {}
for key in files.keys():
  file = base + files[key]['file']
  data[key] = np.genfromtxt(file)

```

```python
for key in data.keys():
  x = files[key]['x']
  all_ys = files[key]['y']

  plt.plot(data[key][:, x], data[key][:,all_ys], label = key)
  plt.legend()
  plt.show()
```
