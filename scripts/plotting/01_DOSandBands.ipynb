```python
import matplotlib.pyplot as plt
from PIL import Image

# Load your images

m1 = '/home/fes33/Documents/GIK - R&D/Personal - Papers and Reports/--Libraries/abinit/Abinitio-Codes/plots/DFT_bands/CsbbBr3.png'
m2 = '/home/fes33/Documents/GIK - R&D/Personal - Papers and Reports/--Libraries/abinit/Abinitio-Codes/plots/DFT_PDOS/CsPbBr3.png'
# m3 = '/home/fes33/Documents/GIK - R&D/Personal - Papers and Reports/--Libraries/abinit/Abinitio-Codes/plots/DFT_DOS/CsPBBr3.png'
m3 = '/home/fes33/Documents/GIK - R&D/Personal - Papers and Reports/--Libraries/abinit/Abinitio-Codes/plots/DFT_PDOS_atom_orb/CsPbBr3.png'

img1 = Image.open(m1)
img2 = Image.open(m2)
img3 = Image.open(m3)


```

```python

# Set the new sizes (width, height)
new_size_img2 = (400, 600)  # Keep original width, set height to 300
new_size_img3 = (400, 600)  # Keep original width, set height to 300

# Resize images 2 and 3 to new dimensions
img2_resized = img2.resize(new_size_img2)
img3_resized = img3.resize(new_size_img3)

# Create a figure with a single row
fig, axes = plt.subplots(1, 3)  # Adjust figsize as needed

# Display images
axes[0].imshow(img1)
axes[0].axis('off')  # Hide axis

axes[1].imshow(img2_resized)
axes[1].axis('off')

axes[2].imshow(img3_resized)
axes[2].axis('off')
plt.subplots_adjust(wspace=-0.6)  # Adjust this value to change the horizontal spacing

# Adjust layout
plt.tight_layout()
plt.savefig('/home/fes33/Documents/GIK - R&D/Personal - Papers and Reports/--Libraries/abinit/Abinitio-Codes/plots/Grids/CsPbBr3.png', format = 'png')
plt.show()
```
